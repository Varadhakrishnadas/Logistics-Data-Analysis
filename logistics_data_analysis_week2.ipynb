```python
import pandas as pd

data = pd.read_csv('/content/logistics_data/ecommerce_logistics_delivery.csv')

print(data.shape)
```

```python
print(data.columns.tolist())
```

```python
missing_values = data.isnull().sum()

print(missing_values)
```

```python
duplicate_count = data.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
print(data.dtypes)
```

```python
print(data['Order_Date'].head())
```

```python
data['Order_Date'] = pd.to_datetime(data['Order_Date'])

print(data['Order_Date'].dtype)
```

```python
print(data.describe())
```

```python
Q1 = data['Order_Value'].quantile(0.25)
Q3 = data['Order_Value'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = data[
    (data['Order_Value'] < lower_bound) |
    (data['Order_Value'] > upper_bound)
]

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of Order Value outliers:", len(outliers))
```

```python
print(outliers['Order_Value'].describe())
print(outliers[['Order_ID', 'Order_Value', 'Order_Quantity', 'Shipping_Cost']].head(10))
```

```python
Q1 = data['Distance_KM'].quantile(0.25)
Q3 = data['Distance_KM'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

distance_outliers = data[
    (data['Distance_KM'] < lower_bound) |
    (data['Distance_KM'] > upper_bound)
]

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of Distance outliers:", len(distance_outliers))
```

```python
print(distance_outliers['Distance_KM'].describe())

print(distance_outliers[
    ['Order_ID', 'Distance_KM', 'Shipping_Method', 'Delivery_Partner', 'Delivery_Delay_Hours']
].head(10))
```

```python
Q1 = data['Traffic_Delay_Minutes'].quantile(0.25)
Q3 = data['Traffic_Delay_Minutes'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

traffic_outliers = data[
    (data['Traffic_Delay_Minutes'] < lower_bound) |
    (data['Traffic_Delay_Minutes'] > upper_bound)
]

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of Traffic Delay outliers:", len(traffic_outliers))
```

```python
print(traffic_outliers['Traffic_Delay_Minutes'].describe())

print(traffic_outliers[
    ['Order_ID', 'Traffic_Delay_Minutes', 'Distance_KM',
     'Shipping_Method', 'Delivery_Delay_Hours']
].head(10))
```

```python
Q1 = data['Delivery_Delay_Hours'].quantile(0.25)
Q3 = data['Delivery_Delay_Hours'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

delivery_delay_outliers = data[
    (data['Delivery_Delay_Hours'] < lower_bound) |
    (data['Delivery_Delay_Hours'] > upper_bound)
]

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of Delivery Delay outliers:", len(delivery_delay_outliers))
```

```python
numeric_columns = [
    'Order_Value',
    'Order_Quantity',
    'Package_Weight_KG',
    'Package_Volume_CM3',
    'Distance_KM',
    'Warehouse_Processing_Hours',
    'Traffic_Delay_Minutes',
    'Weather_Delay_Minutes',
    'Courier_Rating',
    'Courier_Experience_Years',
    'Warehouse_Capacity_Utilization',
    'Estimated_Delivery_Hours',
    'Promised_Delivery_Hours',
    'Delivery_Delay_Hours',
    'Shipping_Cost',
    'Logistics_Cost',
    'Customer_Satisfaction_Score'
]

print(data[numeric_columns].describe().T)
```

```python
from sklearn.preprocessing import MinMaxScaler

data_normalized = data.copy()

scaler = MinMaxScaler()

data_normalized[numeric_columns] = scaler.fit_transform(
    data[numeric_columns]
)

print(data_normalized[numeric_columns].head())
```

```python
print(data_normalized[numeric_columns].min())
print(data_normalized[numeric_columns].max())
```

```python
data_normalized.to_csv(
    '/content/logistics_data/ecommerce_logistics_delivery_cleaned.csv',
    index=False
)

print("Cleaned dataset saved successfully.")
```
