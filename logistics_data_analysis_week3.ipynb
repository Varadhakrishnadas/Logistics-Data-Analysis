```python
Q1 = data['Order_Value'].quantile(0.25)
Q3 = data['Order_Value'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = data[
    (data['Order_Value'] < lower_bound) |
    (data['Order_Value'] > upper_bound)
]

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of Order Value outliers:", len(outliers))
```

```python
print(outliers['Order_Value'].describe())
print(outliers[['Order_ID', 'Order_Value', 'Order_Quantity', 'Shipping_Cost']].head(10))
```

```python
Q1 = data['Distance_KM'].quantile(0.25)
Q3 = data['Distance_KM'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

distance_outliers = data[
    (data['Distance_KM'] < lower_bound) |
    (data['Distance_KM'] > upper_bound)
]

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of Distance outliers:", len(distance_outliers))
```

```python
print(distance_outliers['Distance_KM'].describe())

print(distance_outliers[
    ['Order_ID', 'Distance_KM', 'Shipping_Method', 'Delivery_Partner', 'Delivery_Delay_Hours']
].head(10))
```

```python
import pandas as pd

data = pd.read_csv(
    '/content/logistics_data/ecommerce_logistics_delivery_cleaned.csv'
)

print("Dataset shape:", data.shape)
print(data.head())
```

```python
# Central tendency analysis

key_columns = [
    'Order_Value',
    'Order_Quantity',
    'Distance_KM',
    'Warehouse_Processing_Hours',
    'Traffic_Delay_Minutes',
    'Weather_Delay_Minutes',
    'Delivery_Delay_Hours',
    'Shipping_Cost',
    'Logistics_Cost',
    'Customer_Satisfaction_Score'
]

central_tendency = data[key_columns].agg(['mean', 'median'])

print(central_tendency.round(3))
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Delivery Delay
axes[0].hist(data['Delivery_Delay_Hours'], bins=30)
axes[0].set_title('Distribution of Delivery Delay')
axes[0].set_xlabel('Normalized Delivery Delay')
axes[0].set_ylabel('Number of Orders')

# Logistics Cost
axes[1].hist(data['Logistics_Cost'], bins=30)
axes[1].set_title('Distribution of Logistics Cost')
axes[1].set_xlabel('Normalized Logistics Cost')
axes[1].set_ylabel('Number of Orders')

# Customer Satisfaction
axes[2].hist(data['Customer_Satisfaction_Score'], bins=30)
axes[2].set_title('Distribution of Customer Satisfaction')
axes[2].set_xlabel('Normalized Satisfaction Score')
axes[2].set_ylabel('Number of Orders')

plt.tight_layout()
plt.show()
```

```python
# Correlation analysis

correlation_columns = [
    'Distance_KM',
    'Warehouse_Processing_Hours',
    'Traffic_Delay_Minutes',
    'Weather_Delay_Minutes',
    'Estimated_Delivery_Hours',
    'Promised_Delivery_Hours',
    'Delivery_Delay_Hours',
    'Shipping_Cost',
    'Logistics_Cost',
    'Customer_Satisfaction_Score'
]

correlation_matrix = data[correlation_columns].corr()

print(correlation_matrix.round(2))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))

plt.imshow(correlation_matrix, aspect='auto')
plt.colorbar(label='Correlation')

plt.xticks(
    range(len(correlation_columns)),
    correlation_columns,
    rotation=45,
    ha='right'
)

plt.yticks(
    range(len(correlation_columns)),
    correlation_columns
)

plt.title('Correlation Heatmap of Key Logistics Variables')

plt.tight_layout()
plt.show()
```

```python
shipping_analysis = data.groupby('Shipping_Method').agg(
    Orders=('Order_ID', 'count'),
    Avg_Delivery_Delay=('Delivery_Delay_Hours', 'mean'),
    Avg_Logistics_Cost=('Logistics_Cost', 'mean'),
    Avg_Satisfaction=('Customer_Satisfaction_Score', 'mean'),
    On_Time_Rate=('On_Time_Delivery_Flag', 'mean')
).sort_values('Avg_Delivery_Delay', ascending=False)

print(shipping_analysis.round(3))
```

```python
import matplotlib.pyplot as plt

shipping_plot = shipping_analysis.reset_index()

plt.figure(figsize=(9, 5))

plt.bar(
    shipping_plot['Shipping_Method'],
    shipping_plot['On_Time_Rate']
)

plt.title('On-Time Delivery Rate by Shipping Method')
plt.xlabel('Shipping Method')
plt.ylabel('On-Time Delivery Rate')

plt.ylim(0, 0.6)
plt.tight_layout()
plt.show()
```

```python
partner_analysis = data.groupby('Delivery_Partner').agg(
    Orders=('Order_ID', 'count'),
    Avg_Delivery_Delay=('Delivery_Delay_Hours', 'mean'),
    Avg_Logistics_Cost=('Logistics_Cost', 'mean'),
    Avg_Satisfaction=('Customer_Satisfaction_Score', 'mean'),
    On_Time_Rate=('On_Time_Delivery_Flag', 'mean')
).sort_values('On_Time_Rate', ascending=False)

print(partner_analysis.round(3))
```

```python
monthly_analysis = data.groupby('Month').agg(
    Orders=('Order_ID', 'count'),
    Avg_Delivery_Delay=('Delivery_Delay_Hours', 'mean'),
    Avg_Logistics_Cost=('Logistics_Cost', 'mean'),
    Avg_Satisfaction=('Customer_Satisfaction_Score', 'mean'),
    On_Time_Rate=('On_Time_Delivery_Flag', 'mean')
)

print(monthly_analysis.round(3))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.plot(
    monthly_analysis.index,
    monthly_analysis['On_Time_Rate'],
    marker='o'
)

plt.title('Monthly On-Time Delivery Rate')
plt.xlabel('Month')
plt.ylabel('On-Time Delivery Rate')

plt.xticks(range(1, 13))
plt.ylim(0, 0.4)

plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
